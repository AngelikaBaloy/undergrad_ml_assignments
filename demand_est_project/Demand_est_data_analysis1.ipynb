2. Plot the following over time by brand:

*   average price


```python
sns.lineplot(data=df, x='year', y='avg_price', hue='brand')
```

* average rating

```python
sns.lineplot(data=df, x='year', y='avg_rating', hue='brand')
```

* brand market share

```python
sns.lineplot(data=df, x='year', y='brand_share', hue='brand')
```

3. Summarize the product characteristics:
* Which features are common?

Across years and brands, compact and oven style features are the most common among these air-fryers. Compact is the most common overall.

```python
feature_list = ['compact_share', 'dual_basket_share','oven_style_share', 'rotisserie_share', 'window_share']
df[feature_list].describe()
```

* Which features are rare?

Across brands and years, dual-basket, rotisserie, and window features are the most rare among these air-fryers.

* Are there brands that seem to specialize in different product types?

Yes, each brand seems to focus on a couple of product types rather than selling a range of air-fryers with different features. An exception to this is the Chefman brand, which has decently sized shares across all product characteristics (dual-basket share being the lowest overall).

```python
df.groupby('brand')[feature_list].mean()
```

4. Write a short paragraph describing the market. Which brands are expensive? Which brands have large shares? Does the market look stable over time?

```python
# Avg price by brand
df.groupby('brand')['avg_price'].mean().sort_values(ascending=False)

# cuisinart, oster, and ninja are the three most expensive overall
```

```python
# Shares by brand (across years)
df.groupby('brand')['brand_share'].mean().sort_values(ascending=False)

# ninja seems to have the highest purchase share across years, followed by instant pot and gowise
```
