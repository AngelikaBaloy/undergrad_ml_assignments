## Demand Model

* Using this air fryer data, we created a demand model to infer markups and unit costs.
* Multinomial logistic regression, estimated by linear regression
* Average price, average rating, brand, year, and product characteristic shares (compact, dual basket, oven-style, rotisserie, and window) were used to fit the model and predict log brand share.  
* Our model's R-squared was 0.76, which indicates that our selection of independent variables successfully predict our dependent variable, log brand share.

## Estimated Price Coefficient
* The estimated price coefficient was -0.038.
* A negative price coefficient indicates that price increases are associated with log brand share decreases.
* Choosing to increase prices through markups can negatively impact a business.

## Product Features
* Most brands in this data set choose to specialize in certain products, likely due to product demand or to fill certain gaps/niches in the market.
  *  Compact and oven-style features are the most common
  * Dual-basket, rotisserie, and window features are less common among these air fryers

* Most brands stick to their set of product features, but a few brands have dramatic increases or decreases in certain feature shares.
* A single brand does not dominate all product types.

## Estimated Unit Costs, Markups, and Share-Weighted Profits

* Unit costs
  * Calculated by dividing brand share by our demand slope then adding average price
  * Brands vary quite a lot when it comes to unit costs.
  * Cuisinart (194.90), Oster (161.80), and Ninja (112.49) have the highest unit costs.
  * Price and unit cost have a positive linear relationship: Products that cost more to make are priced higher
  * Product rating has no relationship to unit cost

* Markups
  * Calculated by subtracting estimated unit cost from average price
  * Although unit costs vary, markups across all brands range from 27 to 32 dollars

* Share-weighted Profits
  * Calculated by multiplying brand share and markup
  * Ninja, Instant Pot, and Gowise USA have the highest profits overall
  * Average profits within a single brand can vary
