```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
# loading in data
df = pd.read_csv('undergrad_ml_assignments/demand_est_project/air_fryers_clean_brand_year.csv')
df.columns
```

Use pd.get_dummies(..., drop_first=True) for brand and year dummies. The dropped brand and dropped year become the reference categories, so all dummy coefficients are interpreted relative to those omitted categories.

```python
# dummies

brand_dummies = pd.get_dummies(df['brand'], drop_first=True, dtype=int)
year_dummies = pd.get_dummies(df['year'].astype(str), drop_first=True, dtype=int)
# 2019 and chefman dropped
```

1. What is the estimated price coefficient, B^price?
* The estimated price coefficient is -0.0377.

2. Is it negative? Why is that important?
* B^price is negative, which means that as price increases, log brand share decreases. This suggests that increases in price negatively impact brands' ability to sell air fryers, and that markups are bad for business.

3. Which product features are associated with higher demand?
* Compact, oven-style, and window features are associated with higher demand. Window, specifically, has the highest coefficient.

4. Which brand dummy coefficients are largest? Remember that these are interpreted relative to the dropped brand.
* The top brand dummy coefficents are cuisinart (6.422), ninja (5.839), and instant pot (4.626). This means that, relative to chefman, these are the brands that have the most market share.


5. Which year dummy coefficients are largest? Remember that these are interpreted relative to the dropped year.
* The year dummies with the largest coefficents are 2020 and 2022, though 2020's coefficient is positive and 2022's negative. 2020 and 2021 have positive coefficients. This means that, relative to 2019, 2020 had higher log brand shares overall. 2022 and 2023 were less successful compared to 2019.


6. What is the model's R^2?
* The R^2 is 0.763.

```python
y = df['log_brand_share']

feature_list = ['compact_share', 'dual_basket_share','oven_style_share', 'rotisserie_share', 'window_share']

X = pd.concat(
    [df[["avg_price", "avg_rating"] + feature_list], brand_dummies, year_dummies],
    axis=1,)

model = LinearRegression()
model.fit(X, y)

predicted_log_share = model.predict(X)
r2 = r2_score(y, predicted_log_share)

coef_table = pd.DataFrame({
    "feature": X.columns,
    "coefficient": model.coef_})

print("R-squared:", r2)
coef_table
```
